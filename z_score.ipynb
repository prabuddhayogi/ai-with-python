```python
import numpy as np
from scipy.stats import zscore
```

```python
data = np.array([10,20,30,40,700,800,900])

# calculate mean and standard deviation
mean = np.mean(data)
std_dv = np.std(data)

# Print results
print(f"Data: {data}")
print(f"Mean (u): {mean: .2f}")
print(f"Standard Deviation (o): {std_dv: .2f}")

# z score with formula : x - mean / std_dev
z_score_manual = [(x - mean) / std_dv for x in data]
print(f"\nZ-scores (Manual calculation): {np.round(z_score_manual, 3)}")

# z-score using scipy.stats zscore
z_score_scipy = zscore(data)
print(f"Z-score (scipy.stats.zscore): {np.round(z_score_scipy, 3)}")
```
