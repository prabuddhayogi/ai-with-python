```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
square_size = 50
board_size = 4
```

Two parameters are defined in this step:

square_size Represents the size of each square in pixels. In this case, each square is 50×50 pixels.

board_size Represents the number of rows and columns in the chessboard. A standard chessboard has 8 rows and 8 columns.

These parameters determine the final dimensions of the generated chessboard image.

```python
chessboard = np.zeros((board_size*square_size, board_size*square_size, 3), dtype=np.uint8)
```

An empty image matrix is created using NumPy.

The matrix dimensions are calculated as:

(height, width, color_channels)

Since the chessboard contains 8 squares per side and each square is 50 pixels, the total image size becomes:

400 × 400 pixels

The third dimension represents three color channels (RGB).

The data type uint8 ensures that pixel values range from 0 to 255, which is the standard range for image intensity values.

```python
for row in range(board_size):
    for col in range(board_size):

        if (row + col) % 2 == 0:
            chessboard[row*square_size:(row+1)*square_size,
                       col*square_size:(col+1)*square_size] = [255,255,255]
        else:
            chessboard[row*square_size:(row+1)*square_size,
                       col*square_size:(col+1)*square_size] = [0,0,0]
```

Nested loops are used to iterate through each square in the chessboard.

The condition:

(row + col) % 2

determines whether the square should be black or white.

If the sum of row and column indices is even, the square is filled with white color (255,255,255). If the sum is odd, the square is filled with black color (0,0,0).

NumPy slicing is used to fill the corresponding pixel region for each square.

This process creates the alternating pattern of a standard chessboard.

```python
plt.imshow(chessboard)
plt.title("Original Chessboard")
plt.axis("off")
plt.show()
```

The generated chessboard image is displayed using Matplotlib.

The imshow() function visualizes the image matrix, while the title provides a label for the displayed output.

The axis is turned off to remove coordinate labels and improve visual clarity.

This step verifies that the chessboard pattern has been generated correctly.

```python
row = int(input("Enter row (0-3): "))
col = int(input("Enter column (0-3): "))

print("Choose color:")
print("1 = Red")
print("2 = Green")
print("3 = Blue")

choice = int(input("Enter color number: "))

if choice == 1:
    color = [255,0,0]
elif choice == 2:
    color = [0,255,0]
elif choice == 3:
    color = [0,0,255]
else:
    color = [255,255,0]

chessboard[row*square_size:(row+1)*square_size,
           col*square_size:(col+1)*square_size] = color
plt.imshow(chessboard)
plt.title("Updated Chessboard")
plt.axis("off")
plt.show()
```

In this step, the user is asked to provide the coordinates of the square whose color needs to be modified.

The row and column values range from 0 to 7, corresponding to the 8×8 structure of the chessboard.

The user is given the option to select a color that will be applied to the chosen square.

Each color corresponds to specific RGB values:

Red → [255,0,0]

Green → [0,255,0]

Blue → [0,0,255]

If the user enters a value outside the defined options, a default color (yellow) is assigned.

This step demonstrates how RGB values control pixel colors in digital images.

The selected square is modified by changing the pixel values within the corresponding region of the image matrix.

The coordinates are converted into pixel ranges using the square_size parameter.

For example:

row = 2 col = 3

The affected region becomes:

pixels 100–150 (row)

pixels 150–200 (column)

All pixels within this region are assigned the new RGB color value.

The modified chessboard image is displayed.

The selected square now appears with the new color chosen by the user.

This confirms that the pixel matrix manipulation was successful.
