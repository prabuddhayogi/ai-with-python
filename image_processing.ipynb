Convet image to grayscale 'L'

```python
## Open and show an image, once again. ##
from PIL import Image
import matplotlib.pyplot as plt
img=Image.open('/content/drive/MyDrive/AI-python/lecture32/horse.jpg')

## Convert it to a grayscale image:
## L means grapscale mode，you can try RGB, F mode;
img_convert = img.convert('L')

## Tip: If having error in this line, you may need upgrade matplotlib like this:
## pip install matplotlib --upgrade

##
plt.imshow(img_convert)
plt.show()

# The result will be showed below.
```

Convert image to black and white

```python
WHITE, BLACK = 200, 10   # define gray constant value

img_point = img.point(lambda x: WHITE if x > 128 else BLACK)  # judge each pixel

img_binary = img_point.convert('1')  # convert to binary image

##
plt.imshow(img_binary)
plt.show()

# The result will be showed below.
```

```python
### read a image then display it###
from PIL import Image
import matplotlib.pyplot as plt

## Open the image
img_noise=Image.open('/content/drive/MyDrive/AI-python/lecture32/horse.jpg')

##
plt.figure()
plt.subplot(1,2,1)
plt.imshow(img_noise)
plt.subplot(1,2,2)

## MedianFilter, to remove noise
img_filter = img_noise.filter(ImageFilter.MedianFilter(5))

##
plt.imshow(img_filter)
plt.show()

# The result will be showed below.
```

```python
import cv2
import matplotlib.pyplot as plt

# 1. Load the image (Verify the path is correct!)
img = cv2.imread('/content/drive/MyDrive/AI-python/lecture32/horse.jpg')

if img is None:
    print("Error: Could not load image. Check the file path.")
else:
    # 2. Convert to Grayscale
    gray_img = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    plt.figure(figsize=(10, 5))

    # 3. First Subplot: The Image
    plt.subplot(1, 2, 1)
    plt.title("Grayscale Image")
    plt.imshow(gray_img, cmap='gray')

    # 4. Second Subplot: The Histogram
    plt.subplot(1, 2, 2)
    plt.title("Histogram")
    plt.hist(gray_img.ravel(), 256, [0, 256])

    # 5. Essential: Show the final result
    plt.show()
```
