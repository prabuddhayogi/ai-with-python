```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt
```

```python
df=pd.read_csv("/content/drive/MyDrive/AI-python/lecture11/train.csv")
df
```

```python
df.describe()
```

```python
df.drop(['SibSp','Embarked','Parch','Ticket','Cabin','Fare'],axis=1,inplace=True)
```

```python
df
```

```python
df.isnull()
```

```python
df.fillna("missing")
```

```python
female = df.loc[df.Sex == 'female']["Survived"]
female_survival_rate = sum(female)/len(female) * 100
female_survival_rate
```

```python
male = df.loc[df.Sex == 'male']["Survived"]
male_survival_rate = sum(male)/len(male) * 100
male_survival_rate
```

```python
plt.figure(figsize=(6,6))
plt.pie((male_survival_rate,female_survival_rate), labels=['Male', 'Female'], autopct='%1.1f%%', startangle=90)
plt.title("Survival rate according to Gender")
plt.axis('equal')
plt.show()
```

```python
pclass = np.unique(df['Pclass'])
pclass
```

```python
summary = df.groupby("Pclass").count()
summary
```

```python
# Create bar chart
plt.bar(pclass, summary.PassengerId.values)
plt.xlabel('Passenger Count')
plt.ylabel('Pclss')
plt.title('Pclass passenger counts')
plt.show()
```
