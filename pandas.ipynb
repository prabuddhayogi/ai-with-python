```python
# Random seed
np.random.seed(42)
# No of rows
num_rows= 100
ids=np.arange(1,num_rows+1)
# Column name
names=np.random.choice(['Alice','Bob','Charlie','David','Eva','Frank','Gaby','Hannah'], size=num_rows)
# Column gender
genders=np.random.choice(['Male','Female'],size=num_rows)
# Height
heights=np.random.randint(140, 201, size=num_rows)
# Weight
weights = np.random.randint(100, 200, size=num_rows)

# Food type
lunch_options= ['Pizza', 'Sandwich', 'Salad', 'Burger', 'Pasta', 'Sushi', 'Rice', 'Beans']
breakfast_options= ['Cereal','Pancakes','Eggs','Bread', 'Milk','Fruit', 'Yogurt']
dinner_options = ['Steak', 'Fish', 'Rice & Veg', 'Pizza','Pasta', 'Burger', 'Milk','Salad']

lunch=np.random.choice(lunch_options, size=num_rows)
breakfast=np.random.choice(breakfast_options, size=num_rows)
dinner=np.random.choice(dinner_options, size=num_rows)

# health habits
exercise=np.random.choice([True,False], size=num_rows)
# any disease
disease=np.random.choice([True,False],size=num_rows)

#Create DataFrame
df = pd.DataFrame({
    'ID': ids,
    'Name':names,
    'Gender':genders,
    'Height': heights,
    'Weight': weights,
    'Lunch': lunch,
    'Breakfast': breakfast,
    'Dinner': dinner,
    'Exercise': exercise,
    'Disease': disease
})

# Save to CSV
df.to_csv("/content/drive/MyDrive/AI-python/lecture8/random_health_dat.csv", index=False)
```
